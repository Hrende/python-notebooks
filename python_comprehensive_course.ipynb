```python
#Dna dizisinden baz sayılarını bulma
Dna_dizisi=["AGAAAGGGAAAGGGGTTTCCATTACATTTTATACATGCTA"]

for Dna in Dna_dizisi:
    if "A" in Dna:
        print(f"Adenin bazından {len(Dna)} tane vardır.") #baz kopozisyonu ilkelerini buraya şart olarak ekleyebilirsin. baz sayısına göre kaç tane 
#hidrojen sayısı olduğunu ekle ve bunun Tm değerini nasıl eklediğini belirten bir mesajı ekrana yazdır.

for dna in Dna_dizisi:
    
    A=dna.count("A")
    T=dna.count("T")
    G=dna.count("G")
    C=dna.count("C")

    print(f"\nA sayısı:{A}")
    print(f"T sayısı:{T}")
    print(f"G sayısı:{G}")
    print(f"C sayısı:{C}")

    hidrojen=(A*2)+(T*2)+(G*3)*(C*3)
    print(f"Toplam hidrojen sayısı: {hidrojen}")

    # Tm (melting temperature)
    # Basit Wallace kuralı: Tm = 2*(A+T) + 4*(G+C)
    Tm=2*(A+T) + 4*(G+C)
    print(f"\nTm değeri:{Tm} derecedir.")
```

```python
prompt = "\nBana bir şey söyle, sana onu tekrar edeyim:"
prompt += "\nProgramı sona erdirmek için 'quit' giriniz."

message = ""  
#kullanıcının girdiklerinin kaydını tutmak için message değişkeni oluşturldu
#message boş bir dizgi olarak tanımlıyoruz böylece python nın while satırına ilk kez ulaştığında kontrol edecek birşeyleri oluyor.
#eğer başta python karşılaştıracak birşeyi yoksa program çalışmaya devam etmez
#Bu boş dizgi python un while döngüsünün çalışmasını sağlayan karşılaştırma işlemini gerçekleştirmesini sağlayacaktır.

while True: #daha temiz ve güvenilir.
    message = input(prompt).strip().lower()
    if message == 'quit':
        break
    print(message) # normalde program bu değildi ama diğeri çalışmadı kitapta o var SONRA HATAYI BULDUN ÇALIŞTI.
```

```python
def make_pizza(*toppings):
    """istenilen malzemelerin listesini ekrana yazdırır."""
    print("\nAşağıdaki malzemelerden bir pizza yapılıyor:")
    for topping in toppings:
        print(f"-{topping}")
        

make_pizza("pepperoni")
make_pizza("mantar","yeşil biber","fazladan peynir")
```

```python
#keyfi argüman fonksiyon tanımında konumsal ve anahtar kelime argümanından sonra gelir
#python önce konumsal ve anahtar kelime argümanları eşleştirir ve geri kalan argümanları son parametrede toplar.
def make_pizza(size,*toppings):
    """istenilen malzemelerin listesini ekrana yazdırır."""
    print(f"\nAşağıdaki malzemelerden bir {size} inçlik pizza yapılıyor:")
    for topping in toppings:
        print(f"-{topping}")
        

make_pizza(16,"pepperoni")
make_pizza(18,"mantar","yeşil biber","fazladan peynir")

#Not: genel parametre ismi olan *args'ı sık sık göreceksiniz. Bu parametre bunun gibi keyfi konumsal argümanları alır.
```

```python
#keyfi anahtar kelime argümanları kullanmak
#keyfi anahtar kelime argüman kabul etmek isteyeceksiniz ancak fonks. hangi türde bilginin gönderileceğini önceden bilmiyeceksiniz
#bu durumda çağıran ifadenin sağladığı kadar anahtar-değer çiftlerini kabul eden fonks. yazabiliriz.

def build_profile(first, last, **user_info):
    #parametreden önce iki yıldız user_info boş SÖZLÜĞÜNÜ oluşturur ve aldığı her isim-değer çiftini bu sözlüğe paketlemesine neden olur.
    """kullanıcının hakkında bildiğimiz her şeyi içeren bir sözlük oluştur."""
    user_info["first_name"]=first
    user_info["last_name"]=last
    return user_info #döndürülen değerin atanabileceği bir değişken temin edilmesi gerekiyor "user_profile".

user_profile=build_profile("albert","einstein",location="princeton",field="phycics")#fonks, fonksiyon çağrısında kaç tane ilave anahtar-değer çifti sağlanırsa sağlansın iş görürü.
print(user_profile)

#Not:özel olmayan anahtar kelime argümanlarını alan parametre ismi **kwargs'ı sık sık göreceksiniz.
```

```python
#10.3.1 ZeroDivisionError özel durumunu ele almak
print(5/0)

#python bu tür bir özel durum nesnesini yapmasını istediğimiz şeyi yapamadığı bir duruma yanıt olarak oluşturur.
#Bu olduğunda python programı durdurur ve oluşan özel durumu bize geri bildirir. Bu bilgiyi programımızı değiştimek için kullanabiliriz.
#Bu tür bir hata oluştuğunda python'a ne yapması gerektiğini söyleyeceğiz.Böylece bu özel durum tekrar olursa hazırlıklı olacağız
```

```python
#10.3.2 try-except blokları kullanmak
#Bir hatanın oluşabileceğini düşündüğünüzde oluşabilecek özel durumu ele almak için try-except bloğu yazabilirsiniz.

try:
    print(5/0)
except ZeroDivisionError:
    print("Sıfıra bölemezsiniz")

#Hataya neden olabilecek ifade try bloğuna yerleştirilir.Eğer try bloğundaki kod çalışırsa except bloğunu atlar ama hata varsa.
#hatayla eşlen bir except bloğu arar ve bloktaki kodu çalıştırır.
```

```python
#Çökmeleri engellemek için özel durumları kullanmak
print("Bana iki sayı verin onları böleyim.")
print("Çıkmak için 'q' giriniz.")

while True:
    first_number = input("\nİlk sayı:")
    if first_number == "q":
        break
    second_number = input("İkinci sayı:")
    if second_number == "q":
        break
    answer = int(first_number)/int(second_number)
    print(answer)
#Bu program hataları ele almak için hiçbir şey yapmıyor bu sebeple sıfıra bölmesi istenince çökmesine sebep oldu.
#oluşan hatanın traceback ini göstermek istenen birşey değildir. teknik olamayana kafa karışıklığı ve kötü niyetliye açık verirsin.
```

```python
#10.3.3 else bloğu
#oluşan hata bölme işlemini yapan satırda dolayısıyla try-except bloğu buraya konulur.
#Başarılı bir şekilde çalışan try bloğuna bağlı herhangi bir kod else bloğuna gider.

print("Bana iki sayı verin onları böleyim.")
print("Çıkmak için 'q' giriniz.")

while True:
    first_number = input("\nİlk sayı:")
    if first_number == "q":
        print("Çıkış yapılıyor...")
        break
    second_number = input("İkinci sayı:")
    if second_number == "q":
        print("çıkış yapılıyor...")
        break
    try: #Python'dan bölme işlemini try bloğunda tamamlamaya çalışmasını söylüyoruz.
        answer = int(first_number)/int(second_number)
    except ZeroDivisionError:#Bu özel durum oluştuğuda nasıl yanıt vermesi gerektiğini söylüyoruz.
        print("Sıfıra bölemezsiz!")
    else:#Başarılı olan try bloğuna bağlı olan herhangi bir kod else bloğuna eklenir.
        print(answer)

#Bazen try bloğu başarılı olduğunda çalışması gereken ilave kod olabilir bu kod else bloğunda olmalıdır.
```

```python
import json
def get_stored_username():
    """Eğer mevcut ise saklı kullanıcı ismini al."""
    filename = "username.json"
    try:
        with open(filename) as f:
            username = json.load(f)
    except (FileNotFoundError, json.JSONDecodeError):
        return None
    else:
        return username

def get_new_username():
    """yeni bir kullanıcı ismi iste."""
    username = input("isminizi giriniz:")
    filename = "username.json"
    with open(filename,"w") as f:
        json.dump(username,f)
    return username

def greet_user():
    """kullanıcıyı ismiyle selamla"""
    username = get_stored_username()
    if username:
        print(f"Tekrardan hoşgeldin, {username}")
    else:
        username = get_new_username()
        print(f"Geri geldiğinde seni hatırlayacağız, {username}")
        
greet_user()
```

```python
#11.1 Bir Fonksiyonu test etmek

def get_formetted_name(first,last):
    """düzgün formatlanmış tam bir isim üret"""
    full_name= f"{first} {last}"
    return full_name.title()
#get_formetted_name()'in  çalışıp çalışmadığını kontrol etmek için bu progremı kullanan bir fonksiyon yazalım.
```

```python

print("Çıkmak için herhangi bir anda 'q' giriniz:")
while True:
    first = input("\nLütfen bana bir ad veriniz:")
    if first == "q":
        break
    last = input("Lütfen bana bir soyadı verin:")
    if last == "q":
        break

    formetted_name = get_formetted_name(first,last)
    print(f"Düzgün formatlanmış isim: {formetted_name}")
```
